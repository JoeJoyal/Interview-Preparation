**Problem Statement**
- Given a list of strings, group the words that are anagrams.
- Anagrams are words that contain the same letters in a different order.

**Example:**

**Input:** words = ["eat", "tea", "tan", "ate", "nat", "bat"]

**Output:**

```
[
    ["eat", "tea", "ate"],
    ["tan", "nat"],
    ["bat"]
]

```

**Approach**

The trick is:

- Sort each word.
- Use the sorted word as the dictionary key.
- Words with the same sorted key belong to the same group.

**Ex:**

|Word|Sorted Key|
|----|----------|
|eat|aet|
|tea|aet|
|ate|aet|
|tan|ant|
|nat|ant|
|bat|abt|

**Python Code**

````python
words = ["eat", "tea", "tan", "ate", "nat", "bat"]

groups = {}

for word in words:

    key = "".join(sorted(word))

    if key not in groups:
        groups[key] = []

    groups[key].append(word)

print(list(groups.values()))
````

**Problem Statement**

Given an array of integers and a target value, return the indices of the two numbers whose sum equals the target.

**Assume:**

- Exactly one solution exists.
- You cannot use the same element twice.

**Input**

nums = [2, 7, 11, 15]

target = 9

**Output**
[0, 1]


nums[0] + nums[1]

2 + 7 = 9

**Python code**

````python
nums = [2, 7, 11, 15]
target = 9

seen = {}

for i in range(len(nums)):

    current = nums[i]

    complement = target - current

    if complement in seen:
        print([seen[complement], i])
        break
        
    seen[current] =i
````

**Brute Force Solution**

````python
nums = [2, 7, 11, 15]
target = 9

for i in range(len(nums)):
    for j in range(i + 1, len(nums)):

        if nums[i] + nums[j] == target:
            print([i, j])
````

**Problem - 6**

An LRU (Least Recently Used) Cache stores a limited number of items.

When the cache becomes full:

Keep the recently used items.
Remove the least recently used item.


**Example**

Suppose cache capacity = 3

````python
cache = {}

capacity = 3

# Put 1
cache[1] = "A"

# Put 2
cache[2] = "B"

# Put 3
cache[3] = "C"

print(cache)

# Get 2 (make it recently used)
value = cache.pop(2)
cache[2] = value

print(cache)

# Put 4
if len(cache) == capacity:
    cache.pop(next(iter(cache)))   # Remove least recently used

cache[4] = "D"

print(cache)
````

````python
from collections import OrderedDict

cache = OrderedDict()

capacity = 3

cache[1] = "A"
cache[2] = "B"
cache[3] = "C"

# Access key 2
cache.move_to_end(2)

# Add key 4
if len(cache) == capacity:
    cache.popitem(last=False)   # Remove least recently used

cache[4] = "D"

print(cache)
````

**Problem - 7: Give a sentence, count the frequency of each word**

**Input:** text = "python is easy python is powerful"

````python
text = "python is easy python is powerful"

words = text.split()

frequency = {}

for word in words:

    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
````

**Problem - 8: Matrix Transponse, Given a matrix (2D list), return its transponse**

Transponse means: Convert rows into columns and columns into rows

````python
matrix = [
    [1,2,3],
    [4,5,6]
]

transpose = []

for col in range(len(matrix[0])):

    new_row = []

    for row in range(len(matrix)):
        new_row.append(matrix[row][col])

    transpose.append(new_row)

print(transpose)
````
