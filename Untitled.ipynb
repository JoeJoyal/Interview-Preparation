```python
def group_anagram(words):
    groups = {}

    for word in words:

        # Sort the letter to create a key
        key = "".join(sorted(word))

        if key not in groups:
            groups[key] = []

        groups[key].append(word)

    return list(groups.values())
```

```python
# Input
words = ["eat", "tea", "tan", "ate", "nat", "bat"]
```

```python
print(group_anagram(words))
```
