### **1. Build a Wallet System.**

```python
wallet_balance = 1000

while True:
    print("\n===== Wallet System =====")
    print("1. Check Balance")
    print("2. Add Money")
    print("3. Withdraw Money")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        print("Current Balance:", wallet_balance)

    elif choice == "2":
        amount = float(input("Enter amount to add: "))

        if amount > 0:
            wallet_balance += amount
            print("Money added successfully.")
            print("Updated Balance:", wallet_balance)
        else:
            print("Invalid amount.")

    elif choice == "3":
        amount = float(input("Enter amount to withdraw: "))

        if amount <= 0:
            print("Invalid amount.")
        elif amount > wallet_balance:
            print("Insufficient Balance.")
        else:
            wallet_balance -= amount
            print("Withdrawal successful.")
            print("Updated Balance:", wallet_balance)

    elif choice == "4":
        print("Thank you for using the Wallet System.")
        break

    else:
        print("Invalid choice. Please try again.")
```

### **2.Build a Library Management System**

```python
books = ["Python", "Java", "C++", "Data Structures"]
issued_books = []

while True:
    print("\n===== Library Management System =====")
    print("1. View Books")
    print("2. Issue Book")
    print("3. Return Book")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        if len(books) == 0:
            print("No books available.")
        else:
            print("Available Books:")
            for book in books:
                print("-", book)

    elif choice == "2":
        book_name = input("Enter book name to issue: ")

        if book_name in books:
            books.remove(book_name)
            issued_books.append(book_name)
            print(book_name, "issued successfully.")
        else:
            print("Book not available.")

    elif choice == "3":
        book_name = input("Enter book name to return: ")

        if book_name in issued_books:
            issued_books.remove(book_name)
            books.append(book_name)
            print(book_name, "returned successfully.")
        else:
            print("This book was not issued.")

    elif choice == "4":
        print("Thank you for using the Library Management System.")
        break

    else:
        print("Invalid choice. Please try again.")
```

### **3. Build a Student Management System.**

```python
students = {}

while True:
    print("\n===== Student Management System =====")
    print("1. Add Student")
    print("2. View Students")
    print("3. Update Student Marks")
    print("4. Delete Student")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        roll_no = input("Enter Roll Number: ")

        if roll_no in students:
            print("Student already exists.")
        else:
            name = input("Enter Student Name: ")
            marks = int(input("Enter Marks: "))
            students[roll_no] = {"Name": name, "Marks": marks}
            print("Student added successfully.")

    elif choice == "2":
        if len(students) == 0:
            print("No student records found.")
        else:
            print("\nStudent Records")
            print("---------------------------")
            for roll_no, details in students.items():
                print("Roll No:", roll_no)
                print("Name   :", details["Name"])
                print("Marks  :", details["Marks"])
                print("---------------------------")

    elif choice == "3":
        roll_no = input("Enter Roll Number: ")

        if roll_no in students:
            new_marks = int(input("Enter New Marks: "))
            students[roll_no]["Marks"] = new_marks
            print("Marks updated successfully.")
        else:
            print("Student not found.")

    elif choice == "4":
        roll_no = input("Enter Roll Number: ")

        if roll_no in students:
            del students[roll_no]
            print("Student deleted successfully.")
        else:
            print("Student not found.")

    elif choice == "5":
        print("Exiting Student Management System...")
        break

    else:
        print("Invalid choice. Please try again.")
```

### **Build a Parking Lot**

```python
total_slots = 5
available_slots = total_slots
parked_vehicles = []

while True:
    print("\n===== Parking Lot System =====")
    print("1. Park Vehicle")
    print("2. Remove Vehicle")
    print("3. View Parked Vehicles")
    print("4. Check Available Slots")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        if available_slots == 0:
            print("Parking Lot is Full!")
        else:
            vehicle_no = input("Enter Vehicle Number: ")

            if vehicle_no in parked_vehicles:
                print("Vehicle is already parked.")
            else:
                parked_vehicles.append(vehicle_no)
                available_slots -= 1
                print("Vehicle parked successfully.")

    elif choice == "2":
        vehicle_no = input("Enter Vehicle Number: ")

        if vehicle_no in parked_vehicles:
            parked_vehicles.remove(vehicle_no)
            available_slots += 1
            print("Vehicle removed successfully.")
        else:
            print("Vehicle not found.")

    elif choice == "3":
        if len(parked_vehicles) == 0:
            print("No vehicles parked.")
        else:
            print("\nParked Vehicles:")
            for vehicle in parked_vehicles:
                print(vehicle)

    elif choice == "4":
        print("Total Slots     :", total_slots)
        print("Available Slots :", available_slots)
        print("Occupied Slots  :", total_slots - available_slots)

    elif choice == "5":
        print("Thank you for using the Parking Lot System.")
        break

    else:
        print("Invalid choice. Please try again.")
```

### **Build a Todo Application**

```python
tasks = []

while True:
    print("\n===== Todo Application =====")
    print("1. Add Task")
    print("2. View Tasks")
    print("3. Update Task")
    print("4. Delete Task")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        task = input("Enter Task: ")
        tasks.append(task)
        print("Task added successfully.")

    elif choice == "2":
        if len(tasks) == 0:
            print("No tasks available.")
        else:
            print("\nYour Tasks:")
            for i in range(len(tasks)):
                print(i + 1, ".", tasks[i])

    elif choice == "3":
        if len(tasks) == 0:
            print("No tasks available.")
        else:
            print("\nYour Tasks:")
            for i in range(len(tasks)):
                print(i + 1, ".", tasks[i])

            task_no = int(input("Enter task number to update: "))

            if task_no >= 1 and task_no <= len(tasks):
                new_task = input("Enter new task: ")
                tasks[task_no - 1] = new_task
                print("Task updated successfully.")
            else:
                print("Invalid task number.")

    elif choice == "4":
        if len(tasks) == 0:
            print("No tasks available.")
        else:
            print("\nYour Tasks:")
            for i in range(len(tasks)):
                print(i + 1, ".", tasks[i])

            task_no = int(input("Enter task number to delete: "))

            if task_no >= 1 and task_no <= len(tasks):
                tasks.pop(task_no - 1)
                print("Task deleted successfully.")
            else:
                print("Invalid task number.")

    elif choice == "5":
        print("Thank you for using the Todo Application.")
        break

    else:
        print("Invalid choice. Please try again.")
```

### **Add pagination to an API**

```python
students = [
    "Alice", "Bob", "Charlie", "David", "Eva",
    "Frank", "Grace", "Henry", "Ivy", "Jack",
    "Kevin", "Lily", "Mike", "Nancy", "Oscar"
]

page = int(input("Enter Page Number: "))
page_size = int(input("Enter Page Size: "))

start = (page - 1) * page_size
end = start + page_size

if start >= len(students):
    print("No records found.")
else:
    print("\nStudents on Page", page)
    for student in students[start:end]:
        print(student)
```

### **Build a mini search engine.**

```python
documents = [
    "Python is a programming language",
    "Java is used for backend development",
    "C++ is used for system programming",
    "Python supports machine learning",
    "JavaScript is used for web development"
]

while True:
    print("\n===== Mini Search Engine =====")
    print("1. View Documents")
    print("2. Search")
    print("3. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        print("\nAvailable Documents:")
        for i in range(len(documents)):
            print(i + 1, ".", documents[i])

    elif choice == "2":
        keyword = input("Enter keyword to search: ").lower()

        found = False

        print("\nSearch Results:")
        for doc in documents:
            if keyword in doc.lower():
                print(doc)
                found = True

        if found == False:
            print("No matching documents found.")

    elif choice == "3":
        print("Thank you for using the Mini Search Engine.")
        break

    else:
        print("Invalid choice. Please try again.")
```

### **Design a URL Shortener**

```python
url_map = {}
count = 1

while True:
    print("\n===== URL Shortener =====")
    print("1. Shorten URL")
    print("2. Get Original URL")
    print("3. View All URLs")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        long_url = input("Enter Long URL: ")

        short_url = "short" + str(count)
        url_map[short_url] = long_url
        count += 1

        print("Short URL:", short_url)

    elif choice == "2":
        short_url = input("Enter Short URL: ")

        if short_url in url_map:
            print("Original URL:", url_map[short_url])
        else:
            print("Short URL not found.")

    elif choice == "3":
        if len(url_map) == 0:
            print("No URLs available.")
        else:
            print("\nStored URLs:")
            for short_url, long_url in url_map.items():
                print(short_url, "->", long_url)

    elif choice == "4":
        print("Thank you for using the URL Shortener.")
        break

    else:
        print("Invalid choice. Please try again.")
```

### **Design a Cache**

```python
cache = {}

while True:
    print("\n===== Cache System =====")
    print("1. Add to Cache")
    print("2. Get from Cache")
    print("3. Delete from Cache")
    print("4. View Cache")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        key = input("Enter Key: ")
        value = input("Enter Value: ")

        cache[key] = value
        print("Data cached successfully.")

    elif choice == "2":
        key = input("Enter Key: ")

        if key in cache:
            print("Value:", cache[key])
        else:
            print("Key not found in cache.")

    elif choice == "3":
        key = input("Enter Key: ")

        if key in cache:
            del cache[key]
            print("Data deleted successfully.")
        else:
            print("Key not found.")

    elif choice == "4":
        if len(cache) == 0:
            print("Cache is empty.")
        else:
            print("\nCached Data:")
            for key, value in cache.items():
                print(key, "->", value)

    elif choice == "5":
        print("Exiting Cache System...")
        break

    else:
        print("Invalid choice. Please try again.")
```

### **Build a Notification Service**

```python
notifications = []

while True:
    print("\n===== Notification Service =====")
    print("1. Send Notification")
    print("2. View Notifications")
    print("3. Delete Notification")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        message = input("Enter Notification Message: ")
        notifications.append(message)
        print("Notification sent successfully.")

    elif choice == "2":
        if len(notifications) == 0:
            print("No notifications available.")
        else:
            print("\nNotifications:")
            for i in range(len(notifications)):
                print(i + 1, ".", notifications[i])

    elif choice == "3":
        if len(notifications) == 0:
            print("No notifications to delete.")
        else:
            print("\nNotifications:")
            for i in range(len(notifications)):
                print(i + 1, ".", notifications[i])

            index = int(input("Enter notification number to delete: "))

            if index >= 1 and index <= len(notifications):
                notifications.pop(index - 1)
                print("Notification deleted successfully.")
            else:
                print("Invalid notification number.")

    elif choice == "4":
        print("Thank you for using the Notification Service.")
        break

    else:
        print("Invalid choice. Please try again.")
```

### **Design a PromptManager**

```python
prompts = {}

while True:
    print("\n===== Prompt Manager =====")
    print("1. Add Prompt")
    print("2. View Prompts")
    print("3. Update Prompt")
    print("4. Delete Prompt")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        prompt_name = input("Enter Prompt Name: ")

        if prompt_name in prompts:
            print("Prompt already exists.")
        else:
            prompt_text = input("Enter Prompt Text: ")
            prompts[prompt_name] = prompt_text
            print("Prompt added successfully.")

    elif choice == "2":
        if len(prompts) == 0:
            print("No prompts available.")
        else:
            print("\nSaved Prompts:")
            for name, text in prompts.items():
                print("--------------------------")
                print("Name  :", name)
                print("Prompt:", text)

    elif choice == "3":
        prompt_name = input("Enter Prompt Name to Update: ")

        if prompt_name in prompts:
            new_text = input("Enter New Prompt Text: ")
            prompts[prompt_name] = new_text
            print("Prompt updated successfully.")
        else:
            print("Prompt not found.")

    elif choice == "4":
        prompt_name = input("Enter Prompt Name to Delete: ")

        if prompt_name in prompts:
            del prompts[prompt_name]
            print("Prompt deleted successfully.")
        else:
            print("Prompt not found.")

    elif choice == "5":
        print("Exiting Prompt Manager...")
        break

    else:
        print("Invalid choice. Please try again.")
```

## **Build a checkpoint/restart mechanism**

```python
current_progress = ""
checkpoint = ""

while True:
    print("\n===== Checkpoint / Restart System =====")
    print("1. Update Progress")
    print("2. Save Checkpoint")
    print("3. Restart from Checkpoint")
    print("4. View Status")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        current_progress = input("Enter Current Progress: ")
        print("Progress updated successfully.")

    elif choice == "2":
        checkpoint = current_progress
        print("Checkpoint saved successfully.")

    elif choice == "3":
        if checkpoint == "":
            print("No checkpoint available.")
        else:
            current_progress = checkpoint
            print("Restarted from the last checkpoint.")

    elif choice == "4":
        print("\nCurrent Progress :", current_progress)
        print("Saved Checkpoint :", checkpoint)

    elif choice == "5":
        print("Exiting Checkpoint/Restart System...")
        break

    else:
        print("Invalid choice. Please try again.")
```

## **Design a resumable file uploader**

```python
file_name = ""
total_chunks = 0
uploaded_chunks = 0

while True:
    print("\n===== Resumable File Uploader =====")
    print("1. Create Upload")
    print("2. Upload Next Chunk")
    print("3. Resume Upload")
    print("4. View Upload Status")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        file_name = input("Enter File Name: ")
        total_chunks = int(input("Enter Total Chunks: "))
        uploaded_chunks = 0
        print("Upload created successfully.")

    elif choice == "2":
        if file_name == "":
            print("No upload created.")
        elif uploaded_chunks < total_chunks:
            uploaded_chunks += 1
            print("Chunk", uploaded_chunks, "uploaded successfully.")

            if uploaded_chunks == total_chunks:
                print("File uploaded successfully.")
        else:
            print("File upload already completed.")

    elif choice == "3":
        if file_name == "":
            print("No upload available.")
        elif uploaded_chunks == total_chunks:
            print("Upload already completed.")
        else:
            print("Resuming upload from chunk", uploaded_chunks + 1)

    elif choice == "4":
        print("\nFile Name       :", file_name)
        print("Total Chunks    :", total_chunks)
        print("Uploaded Chunks :", uploaded_chunks)
        print("Remaining       :", total_chunks - uploaded_chunks)

    elif choice == "5":
        print("Exiting Resumable File Uploader...")
        break

    else:
        print("Invalid choice.")
```

### **Build an API rate limiter**

```python
users = {}
MAX_REQUESTS = 5

while True:
    print("\n===== API Rate Limiter =====")
    print("1. Make API Request")
    print("2. View Request Count")
    print("3. Reset User Requests")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        user = input("Enter User ID: ")

        if user not in users:
            users[user] = 0

        if users[user] < MAX_REQUESTS:
            users[user] += 1
            print("Request Accepted")
            print("Requests Used:", users[user], "/", MAX_REQUESTS)
        else:
            print("Rate Limit Exceeded!")
            print("Please try again later.")

    elif choice == "2":
        if len(users) == 0:
            print("No users found.")
        else:
            print("\nUser Request Counts:")
            for user, count in users.items():
                print(user, "->", count, "/", MAX_REQUESTS)

    elif choice == "3":
        user = input("Enter User ID: ")

        if user in users:
            users[user] = 0
            print("Request count reset successfully.")
        else:
            print("User not found.")

    elif choice == "4":
        print("Exiting API Rate Limiter...")
        break

    else:
        print("Invalid choice.")
```

### **Design a recommendation engine**

```python
recommendations = {
    "Movies": ["Inception", "Interstellar", "Avengers"],
    "Books": ["Atomic Habits", "Python Crash Course", "The Alchemist"],
    "Music": ["Shape of You", "Believer", "Perfect"],
    "Courses": ["Python", "Java", "Data Structures"]
}

while True:
    print("\n===== Recommendation Engine =====")
    print("1. View Categories")
    print("2. Get Recommendations")
    print("3. Add Recommendation")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        print("\nAvailable Categories:")
        for category in recommendations:
            print("-", category)

    elif choice == "2":
        category = input("Enter Category: ")

        if category in recommendations:
            print("\nRecommended Items:")
            for item in recommendations[category]:
                print("-", item)
        else:
            print("Category not found.")

    elif choice == "3":
        category = input("Enter Category: ")

        if category in recommendations:
            item = input("Enter Item to Add: ")
            recommendations[category].append(item)
            print("Recommendation added successfully.")
        else:
            print("Category not found.")

    elif choice == "4":
        print("Thank you for using the Recommendation Engine.")
        break

    else:
        print("Invalid choice. Please try again.")
```
