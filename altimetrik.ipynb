**Build a mini search engine which: Searches by keyword (title, description)** 

**Features: Sort by relevance machine coding: feature completeness**

````python
searchEngine = [
    {
        "id":1,
        "title":"Python Programming",
        "description":"Learn Python with examples"
    },
    {
        "id":2,
        "title":"Machine Learning",
        "description":"Python for ML and Deep Learning"
    },
    {
        "id":3,
        "title":"Java Basics",
        "description":"Core Java programming"
    }
]
````

````python
keyword = 'python'
results = []

for doc in searchEngine:
    score = 0

    # Check title
    if keyword.lower() in doc['title'].lower():
        score+=2

    # Check description
    if keyword.lower() in doc['description'].lower():
        score+=1

    # Store matching documents
    if score > 0:
        results.append((score, doc))

# Sort by relevance (highest score first)
results.sort(reverse=True)

print("Search Results:")

for score, doc in results:
    print(doc['title'], 'Score:', score)
````

````python
class SearchEngine:

    def __init__(self):
        self.documents = []

    def add_document(self, doc):
        self.documents.append(doc)

    def search(self, keyword):
        keyword = keyword.lower()
        results = []

        for doc in self.documents:

            score = 0

            # Check title
            if keyword in doc['title'].lower():
                score+=2

            # Check description
            if keyword in doc['description'].lower():
                score+=1

            if score > 0:
                results.append((score, doc))

        results.sort(reverse=True)

        return [doc for score, doc in results]
    
````

````python
engine = SearchEngine()
````

**Question-2**

The ticket assistant needs to call three external services for every ticket:
Customer information API
ML classification API
GenAI summarization API
Each API takes approximately 500 ms.
How would you implement this asynchronously so that the calls happen concurrently rather than
sequentially?

**Python: Asynchronous Programming Communication & Articulation: Communication & Articulation**

````python
import asyncio
import time

async def customer_information_api():
    await asyncio.sleep(2)
    return "Customer Info"

async def ml_classification_api():
    await asyncio.sleep(2)
    return "Authentication"

async def summarization_api():
    await asyncio.sleep(2)
    return "Summary"

start_time = time.time()

async def main():
    customer, category, summer = await asyncio.gather(customer_information_api(), ml_classification_api(), summarization_api())
    print(customer)
    print(category)
    print(summer)
await main()

end_time = time.time()

elasped = int(end_time - start_time)
print(f"Time {elasped} second(s)")
````

**Question-3**

You have 100,000 historical tickets with description as input and category as the target.
How would you build a machine-learning model to classify new tickets into categories?
Python: Python ML Frameworks 

**Machine Coding: Feature Completeness Communication & Articulation: Communication**

**1. Collect and Prepare Data**

- Use the ticket description as the input feature (X).
- Use the category as the target label (y).
- Remove duplicate and missing records.
- Clean the text by converting it to lowercase, removing punctuation, and removing extra spaces.

**2. Convert Text to Numerical Features**
Machine learning models cannot understand text directly, so convert the descriptions into numbers using:

- TF-IDF Vectorizer (simple and effective for text classification)
- Alternatively, word embeddings (Word2Vec, GloVe) or transformer embeddings (BERT) for better performance.

**3. Split the Dataset**
Split the 100,000 tickets into:

- Training set: 80%
- Testing set: 20%

This allows the model to learn from one set and be evaluated on unseen data.

**4. Train a Classification Model**
Suitable models include:

- Naive Bayes (fast baseline for text classification)
- Logistic Regression (good accuracy and interpretable)
- Support Vector Machine (SVM) (often performs well on text data)
- For advanced performance, use BERT or another transformer model.

**5. Evaluate the Model**
Measure performance using:

- Accuracy
- Precision
- Recall
-F1-score
- Confusion Matrix

If categories are imbalanced, focus on Precision, Recall, and F1-score rather than accuracy alone.

**6. Predict New Tickets**
For a new ticket:

1. Apply the same text preprocessing.
2. Convert it using the same TF-IDF vectorizer.
3. Pass it to the trained model.
4. The model predicts the most likely category.

**Python ML Frameworks**
- scikit-learn: TF-IDF, Naive Bayes, Logistic Regression, SVM, evaluation metrics.
- TensorFlow or PyTorch: Deep learning models such as LSTM or BERT.
- Hugging Face Transformers: Fine-tuning pretrained BERT models for higher accuracy.

**Overall Workflow**
```
Historical Tickets
        │
        ▼
Data Cleaning
        │
        ▼
Text Vectorization (TF-IDF)
        │
        ▼
Train/Test Split
        │
        ▼
Train Classifier
        │
        ▼
Evaluate Model
        │
        ▼
Save Model
        │
        ▼
New Ticket Description
        │
        ▼

        │
        ▼

```

This approach is scalable for 100,000 tickets and is a standard solution for text classification problems.
