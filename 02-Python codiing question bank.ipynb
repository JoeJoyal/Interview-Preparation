### Find the first non-repeating character in a string

```python
s = "aabbcde"
```

```python
def first_non_repeating(s):
    frequency = {}

    for char in s:
        frequency[char] = frequency.get(char, 0)+1

    for char in s:
        if frequency[char] == 1:
            return char
    return None
```

```python
result = first_non_repeating(s)
```

```python
def are_anagrams(s1, s2):

    # step:1 Normalize the strings
    s1 = s1.lower()
    s2 = s2.lower()

    # step:2 Different lengths cannot be anagrams
    if len(s1) != len(s2):
        return False

    # step:3 Count character in the first string
    freq = {}

    for char in s1:
        freq[char] = freq.get(char, 0)+1

    # step:4 Subtract character counts using the second string
    for char in s2:
        if char not in freq:
            return False
        freq[char] -= 1

        if freq[char] <0:
            return False

    return True

s1 = input().strip()
s2 = input().strip()

if are_anagrams(s1, s2):
    print("Angrams")
else:
    print("Not Anagrams")


print(freq)
```

### 3. Count the frequency of each word in a paragraph

```python
import re

def word_frequency(paragraph):

    frequency ={}

    # Convert to lowercase and extract words
    words = re.findall(r"\b\w+\b", paragraph.lower())

    # count each words
    for word in words:
        frequency[word] = frequency.get(word, 0)+1

    return frequency

paragraph = input()
result = word_frequency(paragraph)

print(result)

for word, count in result.items():
    print(word, count)
```

### Group a list of strings into anagram groups

```python
words = ["eat", "tea", "tan", "ate", "nat", "bat"]
```

```python
groups = {}

for word in words:

    key = "".join(sorted(word))

    if key not in groups:
        groups[key] = []

    groups[key].append(word)

print(list(groups.values()))
```

### Retry a failed API call with exponential backoff

**1. What is exponential backoff?**

Suppose your application calls an external API, but the API temporarily fails because of:

- Network issues
- Server temporarily unavailable (503)
- Too many requests (429)
- Temporary timeout

Instead of continuously calling the API, we wait before retrying.

`Exponential backoff` means increasing the waiting time after each failed attempt.

Example:

|Attempt|Waiting time|
|-------|------------|
|1st failure|1 second|
|2nd failure|2 second|
|3rd failure|4 second|
|4th failure|8 second|
|5th failure|16 second|


**Formula:** `Wait time = Base delay x 2 ^ (attempt - 1)`


Let's calculate:

|attempt|Calculation|Wait|
|-------|-----------|----|
|1	|1 × 2⁰	|1 sec|
|2	|1 × 2¹	|2 sec|
|3	|1 × 2²	|4 sec|
|4	|1 × 2³	|8 sec|
|5	|1 × 2⁴	|16 sec|

```python
import time

max_retries = 5
base_delay = 1

for attempt in range(1, max_retries + 1):
    try:
        print(f"Attempt {attempt}: Calling API....")

        if attempt < 3:
            raise ConnectionError("API temporarily unavailable")

        response = "API call successful"

        print(response)
        break
    except ConnectionError as error:
        print("Error:", error)

        if attempt == max_retries:
            print("Maximum retries reached. API failed.")
            break

        wait_time = base_delay * (2 ** (attempt - 1))

        print(f"Retrying after {wait_time} seconds...")

        time.sleep(wait_time)
            
```
