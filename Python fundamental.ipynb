**Explain list comprehension**

List comprehension is a concise way to create a new list from an existing iterable (such as a list, tuple, or range) using a single line of code. It is often more readable and efficient than using a for loop with append().

**Syntax**

````python
new_list = [expression for item in iterable]
````

With a condition:

````python
new_list = [expression for item in iterable if condition]
````

**Example 1: Create a List of Squares**

Using a for loop:

````python
squares = []

for i in range(1, 6):
    squares.append(i * i)

print(squares)
````

Using list comprehension:

````python
squares = [i * i for i in range(1, 6)]

print(squares)
````

**Example 2: Filter Even Numbers**

````python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = [num for num in numbers if num % 2 == 0]

print(even_numbers)
````

**Example 3: Convert Strings to Uppercase**

````python
names = ["alice", "bob", "charlie"]

uppercase_names = [name.upper() for name in names]

print(uppercase_names)
````

**Advantages of List Comprehension**
- Shorter and cleaner code.
- Easy to read for simple operations.
- Often faster than using a for loop with append().
- Can include filtering conditions.

**Explain Dictionary Comprehension in Python**

Dictionary comprehension is a concise way to create a dictionary using a single line of code. It is similar to list comprehension but creates key-value pairs.

**Syntax**

````python
new_dict = {key_expression: value_expression for item in iterable}
````

With a condition:

````python
new_dict = {key_expression: value_expression for item in iterable if condition}
````

**Example 1: Create a Dictionary of Squares**

Using a for loop:

````python
square_dict = {}

for i in range(1, 6):
    square_dict[i] = i * i

print(square_dict)
````

Using dictionary comprehension:

````python
square_dict = {i: i * i for i in range(1, 6)}

print(square_dict)
````

**Example 2: Filter Even Numbers**

````python
square_dict = {x: x * x for x in range(1, 11) if x % 2 == 0}

print(square_dict)
````

**Example 3: Convert Values to Uppercase**

````python
students = {
    "s1": "alice",
    "s2": "bob",
    "s3": "charlie"
}

upper_students = {key: value.upper() for key, value in students.items()}

print(upper_students)
````

**Advantages of Dictionary Comprehension**
- Produces concise and readable code.
- Faster than creating a dictionary with a loop in many cases.
- Supports filtering with if conditions.
- Useful for transforming existing dictionaries.

**6. What is a Generator in Python?**

A generator is a special type of function that returns one value at a time using the yield keyword instead of return. It generates values lazily, meaning values are produced only when needed, which makes it memory-efficient.

**Syntax**

````python
def generator_name():
    yield value
````

**Example 1: Simple Generator**

````python
def numbers():
    yield 1
    yield 2
    yield 3

gen = numbers()

print(next(gen))
print(next(gen))
print(next(gen))
````

**Example 2: Generator with a Loop**

````python
def count_up_to(n):
    for i in range(1, n + 1):
        yield i

for num in count_up_to(5):
    print(num)
````

**`yield` vs `return`**

Using `return`

````python
def example():
    return 10

print(example())
````

The function ends immediately after `return`.

Using `yield`

````python
def example():
    yield 10
    yield 20

gen = example()

print(next(gen))
print(next(gen))
````

The function pauses after each yield and resumes from that point when called again.

**Advantages of Generators**
- Memory efficient because they generate values one at a time.
- Useful for processing large datasets.
- Faster when you don't need all values at once.
- Can represent infinite sequences.

**Generator Expression**

A generator can also be created using a generator expression, similar to a list comprehension.

````python
squares = (x**2 for x in range(5))

for num in squares:
    print(num)
````

**Generator vs List**

| Feature          | Generator               | List                         |
| ---------------- | ----------------------- | ---------------------------- |
| Syntax           | `(x for x in iterable)` | `[x for x in iterable]`      |
| Memory Usage     | Low (lazy evaluation)   | Higher (stores all elements) |
| Value Generation | One at a time           | All at once                  |
| Reusable         | No (once exhausted)     | Yes                          |


**7. Difference Between Iterator and Generator in Python**

Both iterators and generators are used to access elements one at a time, but they differ in how they are created and used.

| Feature         | Iterator                                                                             | Generator                                              |
| --------------- | ------------------------------------------------------------------------------------ | ------------------------------------------------------ |
| Definition      | An object used to iterate through a collection                                       | A special function that generates values one at a time |
| Creation        | Created using the `iter()` function or by implementing `__iter__()` and `__next__()` | Created using a function with the `yield` keyword      |
| Keyword Used    | `iter()`, `next()`                                                                   | `yield`                                                |
| Code Complexity | More code to implement                                                               | Simple and concise                                     |
| Memory Usage    | Memory efficient                                                                     | More memory efficient due to lazy evaluation           |
| State           | Maintains current iteration state                                                    | Automatically maintains state between `yield` calls    |


**Example 1: Iterator**

````python
numbers = [10, 20, 30]

it = iter(numbers)

print(next(it))
print(next(it))
print(next(it))
````

Here, iter() creates an iterator, and next() retrieves one element at a time.

**Example 2: Generator**

````python
def numbers():
    yield 10
    yield 20
    yield 30

gen = numbers()

print(next(gen))
print(next(gen))
print(next(gen))
````

The generator automatically produces values one by one using `yield`.

**Custom Iterator Example**

````python
class Count:
    def __init__(self, limit):
        self.limit = limit
        self.current = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= self.limit:
            value = self.current
            self.current += 1
            return value
        raise StopIteration

counter = Count(3)

for num in counter:
    print(num)
````

**Equivalent Generator**

````python
def count(limit):
    for i in range(1, limit + 1):
        yield i

for num in count(3):
    print(num)
````

**Key Differences**
  
- Iterator → Uses iter() and next(), or implements __iter__() and __next__().
- Generator → Uses the yield keyword.
- Iterator → More code to implement.
- Generator → Simpler, cleaner, and ideal for lazy evaluation.

**What is yield in Python?**

The yield keyword is used in a function to make it a generator. Unlike return, which ends the function and returns a value, yield returns one value at a time and pauses the function, saving its current state. The function resumes execution from where it left off when the next value is requested.

**Syntax**

````python
def generator():
    yield value
````

**Example 1: Using `yield`**

````python
def numbers():
    yield 1
    yield 2
    yield 3

gen = numbers()

print(next(gen))
print(next(gen))
print(next(gen))
````

The function pauses after each yield and resumes when next() is called.

**Difference Between `yield` and `return`**

| `yield`                     | `return`                      |
| --------------------------- | ----------------------------- |
| Returns one value at a time | Returns a single value        |
| Pauses the function         | Ends the function immediately |
| Used in generators          | Used in normal functions      |
| Saves function state        | Does not save function state  |


**Example**

Using return:

````python
def demo():
    return 10
    return 20

print(demo())
````

**Using `yield:`**

````python
def demo():
    yield 10
    yield 20

gen = demo()

print(next(gen))
print(next(gen))
````

**9. Difference between append() and extend().**

Both `append()` and `extend()` are list methods used to add elements to a list, but they work differently.

| Feature   | `append()`                            | `extend()`                                   |
| --------- | ------------------------------------- | -------------------------------------------- |
| Adds      | A single element                      | Multiple elements from an iterable           |
| Effect    | Adds the entire object as one element | Adds each element individually               |
| Argument  | Any object                            | An iterable (list, tuple, set, string, etc.) |
| List Size | Increases by 1                        | Increases by the number of elements added    |


**Example 1: `append()`**


````python
fruits = ["Apple", "Banana"]

fruits.append("Mango")

print(fruits)
````

**Appending a list:**

````python
fruits = ["Apple", "Banana"]

fruits.append(["Mango", "Orange"])

print(fruits)
````

The entire list is added as a single element.

**Example 2: extend()**

````python
fruits = ["Apple", "Banana"]

fruits.extend(["Mango", "Orange"])

print(fruits)
````

Each element of the iterable is added separately.

**Example with a String**

````python
letters = ['A']

letters.extend("BCD")

print(letters)
````

`extend()` treats the string as an iterable and adds each character individually.

**Interview Answer**
- `append()` adds a single element to the end of a list.
- `extend()` adds all elements of an iterable (such as a list, tuple, or string) to the end of a list.
- `append()` increases the list size by 1, whereas `extend()` increases it by the number of elements in the iterable.

````python
list1 = [1, 2]

list1.append([3, 4])
print(list1)
# Output: [1, 2, [3, 4]]

list2 = [1, 2]

list2.extend([3, 4])
print(list2)
# Output: [1, 2, 3, 4]
````

**10. Difference Between `is` and `==` in Python**

Both `is` and `==` are comparison operators, but they compare different things.

| Feature             | `==`              | `is`                                    |
| ------------------- | ----------------- | --------------------------------------- |
| Compares            | Values (contents) | Object identity (memory location)       |
| Returns `True` when | Values are equal  | Both variables refer to the same object |
| Used for            | Value comparison  | Identity comparison                     |


**Example 1: Using ==**

````python
list1 = [1, 2, 3]
list2 = [1, 2, 3]

print(list1 == list2)
````

**Example 2: Using is**

````python
list1 = [1, 2, 3]
list2 = [1, 2, 3]

print(list1 is list2)
````

Although the values are the same, list1 and list2 are different objects in memory.

**Example 3: Same Object**

````python
list1 = [1, 2, 3]
list2 = list1

print(list1 == list2)
print(list1 is list2)
````

Both variables refer to the same object, so both operators return True.

**Memory Representation**

````python
a = [10, 20]
b = [10, 20]
c = a
````

- a == b → True (same values)
- a is b → False (different objects)
- a is c → True (same object)

**When to Use `is`**

Use `is` when checking for singleton objects such as `None`.

````python
value = None

if value is None:
    print("Value is None")
````

This is the recommended style.

**Interview Answer**
  
- `==` compares the **values** of two objects.
- `is` compares whether two variables refer to the **same object in memory.**

````python
a = [1, 2]
b = [1, 2]
c = a

print(a == b)   # True
print(a is b)   # False
print(a is c)   # True
````

**11. Group Strings into Anagrams**

**Problem Statement:**

Given an array of strings, group the strings that are anagrams of each other.

Anagrams are words that contain the same characters with the same frequencies, but in a different order.

**Example**

**Input**

```
strs = ["eat", "tea", "tan", "ate", "nat", "bat"]
```

**Output**

```
[['eat', 'tea', 'ate'], ['tan', 'nat'], ['bat']]
```

**Approach**

- For each string, sort its characters.
- Use the sorted string as the key in a dictionary.
- Append the original string to the corresponding list.
- Return all the grouped values.

````python
words = ['eat', 'tea', 'tan', 'ate', 'ant', 'bat']

groups = {}

for word in words:
    key = "".join(sorted(word))

    if key not in groups:
        groups[key] = []
    groups[key].append(word)

print(list(groups.values()))

````

````python
from collections import defaultdict

def groupAnagrams(strs):
    groups = defaultdict(list)

    for word in strs:
        key = "".join(sorted(word))
        groups[key].append(word)

    return list(groups.values())

# Example
strs = ["eat", "tea", "tan", "ate", "nat", "bat"]
print(groupAnagrams(strs))
````

**Alternative Solution (Character Count)**

Instead of sorting each string, use the count of each letter as the key.

````python
from collections import defaultdict

def groupAnagrams(strs):
    groups = defaultdict(list)

    for word in strs:
        count = [0] * 26
        for ch in word:
            count[ord(ch) - ord('a')] += 1

        groups[tuple(count)].append(word)

    return list(groups.values())

# Example
strs = ["eat", "tea", "tan", "ate", "nat", "bat"]
print(groupAnagrams(strs))
````

**Time Complexity**

**Sorting Approach**
- Sorting each word: O(K log K), where K is the maximum word length.
- For N words: O(N × K log K).

**Character Count Approach**
- Counting characters: O(K) per word.
- Total: O(N × K).

**Space Complexity**
- Dictionary stores all strings: O(N × K).

**Interview Answer**

Use a dictionary where the key is either:

- the sorted version of each word, or
- a 26-character frequency tuple.

All words with the same key belong to the same anagram group.

**Example:**

````python
from collections import defaultdict

def groupAnagrams(strs):
    groups = defaultdict(list)

    for word in strs:
        groups["".join(sorted(word))].append(word)

    return list(groups.values())
````

This solution has O(N × K log K) time complexity using sorting, or O(N × K) using the character frequency approach.

**12. Find duplicate elements**

Find Duplicate Elements in an Array (Python)
Problem Statement

Given an array of integers, find all elements that appear more than once.

**Example**

**Input:**

```
nums = [1, 2, 3, 2, 4, 5, 1, 6]
```

**Output:**

```
[1, 2]
```

````python
nums = [1, 2, 3, 2, 4, 5, 1, 6]

duplicates = []

for i in range(len(nums)):
    for j in range(i + 1, len(nums)):
        if nums[i] == nums[j]:
            if nums[i] not in duplicates:
                duplicates.append(nums[i])

print("Duplicate elements:", duplicates)
````

````python
nums = [1, 2, 3, 2, 4, 5, 1, 6]

duplicates = []

for num in nums:
    if nums.count(num) > 1 and num not in duplicates:
        duplicates.append(num)

print(duplicates)
````

**12. Find the first non-repeating character**

**Problem Statement**

Given a string, find the first character that appears only once.

**Example**

**Input:**

```
s = "aabbcdde"

```

**Output:**

```
c
```

**Simple Python Solution (Using `count()`)**

````python
s = "aabbcdde"

for ch in s:
    if s.count(ch) == 1:
        print("First non-repeating character:", ch)
        break
````

````python
s = "aabbcdde"

for ch in s:
    if s.count(ch) == 1:
        print(ch)
        break
````

**Time Complexity**

count() scans the string each time, so the overall time complexity is O(n²).

**More Efficient Approach (Using a Dictionary)**

````python
s = "aabbcdde"

freq = {}

# Count frequency
for ch in s:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1

# Find first non-repeating character
for ch in s:
    if freq[ch] == 1:
        print(ch)
        break
````

**Time Complexity: O(n)**

This is the preferred solution in coding interviews because it scans the string only twice.
