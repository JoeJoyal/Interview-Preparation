**Exercise 3 - Set operations**

**Scenario:** Your AI Agent receives duplicate permissions from multiple sources.

```python
role_permissions = {
    "orders:read",
    "orders:cancel",
    "orders:read"
}

type(role_permissions)
```

```python
role_permissions
```

```python
requested_permission = {
    "orders:read",
    "orders:refund"
}

type(requested_permission)

# 1. Display unique role permissions
print(requested_permission)
```

```python
# 2. Find common permissions
print(role_permissions & requested_permission)
```

```python
# 3. Find all permissions from both sets
print(role_permissions | requested_permission)
```

```python
# 4. Find requested permissions the role does not have
print(requested_permission - role_permissions)
```

In a real application, permissions must be checked on the server, set are only a convenient way to represent and compare them.

**Exercise 2:** Count order statuses

**Concept:** Dictionary and loops.

```python
statuses = ["PLACED", "SHIPPED", "PLACED", "DELIVERED", "PLACED"]
```

```python
type(statuses)
```

```python
status_count = {}

for status in statuses:
    status_count[status] = status_count.get(status, 0)+1

print(status_count)
```

```python
status_count = {}

for status in statuses:
    status_count[status] = status_count.get(status, 0)

print(status_count)
```

A shorter alternative uses Python's Counter:

```python
from collections import Counter
```

```python
print(Counter(statuses))
```

**Exercise 4:** Find missing permissions

**Concept:** Sets and set difference.

```python
granted = {"orders:read", "orders:cancel"}
required = {"orders:read", "orders:refund"}
```

```python
required - granted
```

```python
missing_permissions = required - granted
print(missing_permissions)
```

```python
if not missing_permissions:
    print("Access allowed")
else:
    print("Access denied")
```

The required - granted operation finds permissions that are required but not granted. This is a useful way to reason about authorization in an agentic backend. In production, the backend must enforce the permission check before executing the requested action.
